# Comprobación de la entrega de código

Este notebook comprueba archivos y reúne resultados guardados. No ajusta modelos,
no cambia selecciones y no modifica el directorio de resultados. Una comprobación
satisfactoria aquí no equivale a ejecutar de nuevo el paper completo.

Usa una copia nueva del proyecto y configura sus rutas a continuación.


```python
from pathlib import Path
import os
import sys
import subprocess
from datetime import datetime, timezone

try:
    from google.colab import drive
except ImportError:
    pass
else:
    drive.mount("/content/drive")

# Cambia PROJECT_DIR a la carpeta donde descomprimiste esta copia.
PROJECT_DIR = Path(os.environ.get("KNOT_PROJECT_DIR",
    "/content/drive/MyDrive/consensus_hardness_refactored"))
DATA_DIR = Path(os.environ.get("KNOT_DATA_DIR",
    "/content/drive/MyDrive/Colab Notebooks/data_invariants/Invariants"))
OUTPUT_DIR = Path(os.environ.get("KNOT_OUTPUT_DIR",
    str(DATA_DIR / "processed_consensus_hardness/corrected_run_20260819")))
assert (PROJECT_DIR / "scripts/verify_release.py").is_file(), PROJECT_DIR
assert OUTPUT_DIR.is_dir(), OUTPUT_DIR
print("Código:", PROJECT_DIR)
print("Resultados existentes:", OUTPUT_DIR)
```

## Comprobar archivos

La comprobación usa la biblioteca estándar de Python. No necesitas instalar
Sage ni TensorFlow para ejecutar este notebook. Si faltan archivos, se muestran
sus nombres y se conserva el informe para diagnosticarlo.


```python
stamp = datetime.now(timezone.utc).strftime("%Y%m%dT%H%M%SZ")
report = PROJECT_DIR / f"release_check_{stamp}.json"
result = subprocess.run([sys.executable, str(PROJECT_DIR / "scripts/verify_release.py"),
    "--root", str(OUTPUT_DIR), "--data-dir", str(DATA_DIR), "--out", str(report)])
print("Código de salida de la comprobación:", result.returncode)
```

## Recoger las tablas existentes

Continúa aunque el informe indique archivos ausentes: el ZIP documenta lo que
existe y lo que falta. No incluye las matrices y checkpoints grandes. Leerá
las tablas pequeñas, por lo que el tiempo depende principalmente de Drive.


```python
archive = PROJECT_DIR / f"release_evidence_{stamp}.zip"
subprocess.check_call([sys.executable, str(PROJECT_DIR / "scripts/collect_release_evidence.py"),
    "--root", str(OUTPUT_DIR), "--data-dir", str(DATA_DIR), "--out", str(archive)])
print("Archivo para compartir:", archive)
```
